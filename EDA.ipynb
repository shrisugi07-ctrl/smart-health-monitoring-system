```python
# Path to your cleaned dataset
file_path = '/content/drive/MyDrive/Colab Notebooks/clean_waterborne_disease_dataset.csv'

# Optional: output file (if you want to save more cleaned versions)
output_file = '/content/drive/MyDrive/Colab Notebooks/clean_waterborne_disease_dataset_final.csv'

# Chunk size for RAM-safe processing
chunk_size = 50000
```

```python
import pandas as pd

# Preview first 5 rows
sample_df = pd.read_csv(file_path, nrows=5)
print(sample_df.head())
print(sample_df.info())
```

```python
total_rows = 0
missing_values = pd.Series(0)

for chunk in pd.read_csv(file_path, chunksize=chunk_size):
    total_rows += len(chunk)
    missing_values = missing_values.add(chunk.isnull().sum(), fill_value=0)

print(f"Total rows in dataset: {total_rows}")
print("Missing values per column:\n", missing_values)
```

```python
numeric_stats = None

for chunk in pd.read_csv(file_path, chunksize=chunk_size):
    numeric_chunk = chunk.select_dtypes(include='number')
    if numeric_stats is None:
        numeric_stats = numeric_chunk.describe()
    else:
        numeric_stats = numeric_stats.add(numeric_chunk.describe(), fill_value=0)

print("Summary stats for numeric columns:\n", numeric_stats)
```

```python
from collections import Counter

category_counts = {}

for chunk in pd.read_csv(file_path, chunksize=chunk_size):
    for col in chunk.select_dtypes(include='object').columns:
        if col not in category_counts:
            category_counts[col] = Counter()
        category_counts[col].update(chunk[col].dropna())

# Show top 5 values per categorical column
for col, counts in category_counts.items():
    print(f"{col}: {counts.most_common(5)}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import gc  # garbage collector

# Load a safe sample
file_path = '/content/drive/MyDrive/Colab Notebooks/clean_waterborne_disease_dataset.csv'
sample_rows = 50000  # safe size for RAM
sample_df = pd.read_csv(file_path, nrows=sample_rows)

print(f"Loaded sample: {sample_df.shape}")

# Numeric columns histograms (one by one)
numeric_cols = sample_df.select_dtypes(include='number').columns
for col in numeric_cols:
    plt.figure(figsize=(8,4))
    sns.histplot(sample_df[col], kde=True, bins=30, color='skyblue')
    plt.title(f'Distribution of {col}')
    plt.show()
    plt.close()  # free memory
    gc.collect()

# Categorical columns bar plots (top 10 values, one by one)
categorical_cols = sample_df.select_dtypes(include='object').columns
for col in categorical_cols:
    top_values = sample_df[col].value_counts().nlargest(10)
    plt.figure(figsize=(8,4))
    sns.barplot(x=top_values.index, y=top_values.values, palette='viridis')
    plt.title(f'Top 10 categories in {col}')
    plt.xticks(rotation=45, ha='right')
    plt.show()
    plt.close()
    gc.collect()
```

```python
import os

# Folder to save plots
plot_folder = '/content/EDA_Plots'

if not os.path.exists(plot_folder):
    os.makedirs(plot_folder)

print(f"Plots will be saved in: {plot_folder}")
```

```python
# Numeric plots
for col in numeric_cols:
    plt.figure(figsize=(8,4))
    sns.histplot(sample_df[col], kde=True, bins=30, color='skyblue')
    plt.title(f'Distribution of {col}')

    # Save plot to folder
    plt.savefig(os.path.join(plot_folder, f'{col}_hist.png'))
    plt.close()
    gc.collect()

# Categorical plots
for col in categorical_cols:
    top_values = sample_df[col].value_counts().nlargest(10)
    plt.figure(figsize=(8,4))
    sns.barplot(x=top_values.index, y=top_values.values, palette='viridis')
    plt.title(f'Top 10 categories in {col}')
    plt.xticks(rotation=45, ha='right')

    # Save plot to folder
    plt.savefig(os.path.join(plot_folder, f'{col}_bar.png'))
    plt.close()
    gc.collect()
```
